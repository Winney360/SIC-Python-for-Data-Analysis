```python
# How many swaps were executed in the bubble sort process below?
def bubblesort(S):
    n = len(S)
    for i in range(n):
        print(S)
        for j in range(n - 1):
            if S[j] > S[j + 1]:
                S[j], S[j + 1] = S[j + 1], S[j]

S = [50, 30, 40, 10, 20]
bubblesort(S)
print(S)

#6 swaps
```

```python
# How many comparisons were executed in the insertion sort process below?

def insertionsort2(S):
    n = len(S)
    for i in range(1, n):
        print(S)
        x = S[i]
        j = i - 1
        while j >= 0 and S[j] > x:
            S[j + 1] = S[j]
            j -= 1
        S[j + 1] = x

S = [50, 30, 40, 10, 20]
insertionsort2(S)
print(S)

# 9 Comparisons
```
