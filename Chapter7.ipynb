```python
# make a geometric rainbow pattern
import turtle

colors = ['red', 'yellow', 'blue', 'orange', 'green', 'red']

aiden = turtle.Turtle()
aiden.speed(0)  # Optional: speeds up drawing
turtle.bgcolor('black')  # turn background black

# make 36 hexagons, each 10 degrees apart
for n in range(36):
    # make hexagon by repeating 6 times
    for i in range(6):
        aiden.color(colors[i])  # pick color at position i
        aiden.forward(100)
        aiden.left(60)
    # add a turn before the next hexagon
    aiden.right(10)

# get ready to draw 36 circles
aiden.penup()
aiden.color('white')

# repeat 36 times to match the 36 hexagons
for i in range(36):
    aiden.forward(220)
    aiden.pendown()
    aiden.circle(5)
    aiden.penup()
    aiden.backward(220)
    aiden.right(10)

# hide turtle to finish the drawing
aiden.hideturtle()

turtle.done()  # Keeps the window open when completed
```

```python
from turtle import *
from datetime import datetime

def jump(distanz, winkel=0):
    penup()
    right(winkel)
    forward(distanz)
    left(winkel)
    pendown()

def hand(laenge, spitze):
    fd(laenge * 1.15)
    rt(90)
    fd(spitze / 2.0)
    lt(120)
    fd(spitze)
    lt(120)
    fd(spitze)
    lt(120)
    fd(spitze / 2.0)

def make_hand_shape(name, laenge, spitze):
    reset()
    jump(-laenge * 0.15)
    begin_poly()
    hand(laenge, spitze)
    end_poly()
    hand_form = get_poly()
    register_shape(name, hand_form)

def clockface(radius):
    reset()
    pensize(7)
    for i in range(60):
        jump(radius)
        if i % 5 == 0:
            fd(25)
            jump(-radius - 25)
        else:
            dot(3)
            jump(-radius)
        rt(6)

def setup():
    global second_hand, minute_hand, hour_hand, writer
    mode("logo")
    make_hand_shape("second_hand", 125, 25)
    make_hand_shape("minute_hand", 130, 25)
    make_hand_shape("hour_hand", 90, 25)
    clockface(160)

    second_hand = Turtle()
    second_hand.shape("second_hand")
    second_hand.color("gray20", "gray80")

    minute_hand = Turtle()
    minute_hand.shape("minute_hand")
    minute_hand.color("blue1", "red1")

    hour_hand = Turtle()
    hour_hand.shape("hour_hand")
    hour_hand.color("blue1", "red1")

    for h in (second_hand, minute_hand, hour_hand):
        h.shapesize(1, 1, 3)
        h.speed(0)

    writer = Turtle()
    writer.hideturtle()
    writer.penup()

def tick():
    t = datetime.now()
    sec = t.second + t.microsecond * 0.000001
    minute = t.minute + sec / 60.0
    hour = t.hour + minute / 60.0

    tracer(False)
    writer.clear()
    writer.home()
    writer.forward(65)
    writer.write(t.strftime("%A"), align="center", font=("Courier", 14, "bold"))
    writer.back(130)
    writer.write(t.strftime("%b %d %Y"), align="center", font=("Courier", 14, "bold"))

    second_hand.setheading(6 * sec)
    minute_hand.setheading(6 * minute)
    hour_hand.setheading(30 * hour)

    tracer(True)
    ontimer(tick, 100)

def main():
    tracer(False)
    setup()
    tracer(True)
    tick()

if __name__ == "__main__":
    main()
    mainloop()
```

```python
import pandas as pd
import datetime
from datetime import date, datetime, time
import matplotlib.pyplot as plt

import yfinance as yf
start = datetime(2025, 1, 1)
end = datetime(2025, 11, 30)

sec = yf.download(tickers = ['005930.KS'],
                 start = start,
                 end = end,
                 auto_adjust = False)
sec.head
```

```python
close = sec['Adj Close']

ma5 = sec['Adj Close'].rolling(window=5).mean()
ma10 = sec['Adj Close'].rolling(window=10).mean()
ma60 = sec['Adj Close'].rolling(window=60).mean()
ma120 = sec['Adj Close'].rolling(window=120).mean()

plt.figure(figsize = (15,10))

plt.plot(close, label='close', linewidth=4)
plt.plot(ma5, label='5 window')
plt.plot(ma10, label='10 window')
plt.plot(ma60, label='60 window')
plt.plot(ma120, label='120 window')

plt.legend()
plt.title('Rolling window calculations')
plt.xlabel('Date')
plt.ylabel('close price')
```
