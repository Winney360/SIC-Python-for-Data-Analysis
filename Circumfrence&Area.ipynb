```python
pi = 22/7 #the value of pi
radius = float(input("What is the radius of the circle?")) #ask for input of radius

#calculate the circumfrence
circumfrence = 2 * pi * radius
#display the circumfrence
print("The circumfrence of the circle is:", circumfrence)

#calculate the area
area = pi * radius * radius
#display the area
print("The area of the cirle is:" , area)
```
