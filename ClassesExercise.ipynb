```python
class Circle:
    def __init__(self, radius = 0):
        self.__pi = 22/7
        self.__radius = radius
        self.__area = 0
        self.__circumfrence = 0
    def data_in(self):
        self.__radius = float(input("Enter the radius of the circle:"))
    def compute(self):
        self.__area = self.__pi * self.__radius ** 2
        self.__circumfrence = 2 * self.__pi * self.__radius
    def data_out(self):
        print(f"For a circle with radius,{self.__radius} the: ")
        print(f"    area is: {self.__area}")
        print(f"    circumfrence is: {self.__circumfrence}")

def main():
    circle = Circle()
    circle.data_in()
    circle.compute()
    circle.data_out()
    

if __name__  == "__main__":
    main()
        
        
```

```python
class Large:
    def __init__(self, a = 0, b = 0):
        self.__a = a
        self.__b = b
        self.__largest = 0
        
    def data_in(self):
        self.__a = float(input("Enter the first number:"))
        self.__b = float(input("Enter the second number:"))

    def compute(self):
        if self.__a >= self.__b:
            self.__largest = self.__a
        else:
            self.__largest = self.__b
        return self.__largest 

def main():
    large = Large()
    print("We are getting the largest number between 2 numbers:")
    large.data_in()
    print(f"The largest number is : {large.compute()}")
    
if __name__ == "__main__":
    main()
        
    
```

```python
class Small:
    def __init__ (self, a = 0, b = 0, c = 0):
        self.__a = a
        self.__b = b
        self.__c = c
        self.__smallest = 0

    def data_in(self):
        self.__a ,self.__b, self.__c = map(float, input("Enter three numbers separated by a space: ").split())
        
    def compute(self):
        if self.__a <= self.__b and self.__a <= self.__b:
            self.__smallest = self.__a
        elif self.__b <= self.__a and self.__b <= self.__c:
            self.__smallest = self.__b
        else:
            self.__smallest = self.__c
        return self.__smallest

def main():
    small = Small()

    print("We are computing the smallest number among three.")
    print()
    
    small.data_in()
    print()
    
    print(f"The smallest number is : {small.compute()}")
    
if  __name__ == "__main__":
    main()
        
```

```python
class Grade:
    def __init__(self, a = 0, b = 0, c = 0):
        self.__a = a
        self.__b = b
        self.__c = c
        self.__average = 0
        self.__grading = "Incomplete"

    def data_in(self):
        self.__a , self.__b, self.__c = map(float, input("Enter the marks of the three subjects separated by space : ").split())

    def compute_avg(self):
        self.__average = (self.__a + self.__b + self.__c) / 3
        return self.__average

    def compute_grade(self):
        if self.__average >= 70:
            self.__grading = 'A'
        elif self.__average >= 60:
            self.__grading = 'B'
        elif self.__average >= 50:
            self.__grading = 'C'
        elif self.__average >= 40:
            self.__grading = 'D'
        else:
            self.__grading = 'E'
        return self.__grading

def main():
    grade = Grade()
    print("You are about to compute the average of three subject and get the averall grade.")
    print()

    grade.data_in()
    print()

    print(f"The average mark is :{grade.compute_avg():.2f}")
    print()

    print(f"The overall grade is : {grade.compute_grade()}")
    
if __name__ == "__main__":
    main()
        
```
