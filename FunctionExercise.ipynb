```python
#Area of a circle
def fn_area(radius):
    pi = 22/7
    area = pi * radius **2
    return area
#Circumfrence of a circle
def fn_circumfrence(radius):
    pi = 22/7
    circumfrence = 2 * pi * radius 
    return circumfrence
    
def main():
    radius = float(input("Enter the radius:"))
    area = fn_area(radius)
    circumfrence =  fn_circumfrence(radius)
    print(f"The area of a circle with radius {radius}, is {area} , and its circumfrence is {circumfrence}")
if __name__ == "__main__":
    main()


```

```python
# Q 3.(i)
def small(int1,int2,int3):
    if int1 <= int2 and int1 <= int3:
        smallest = int1
    elif int2 <= int3 and int2 <= int1:
        smallest = int2
    else:
        smallest = int3
    return smallest

def main():
    int1,int2,int3 = map(int, input("Enter the 3 integers to get the smallest:").split())
    print(f"The integers are:{int1},{int2},{int3}")
    smallest = small(int1,int2,int3)
    print(f"The smallest is : {smallest}")
if __name__ == "__main__":
    main()
```

```python
# Q.5
def get_big(a,b,c):
    if a >= b and a >= c:
        biggest = a
    elif b >= c and b >= a:
        biggest = b
    else:
        biggest = c
    return biggest

def main():
    a,b,c = map(float, input("Enter the floating numbers separated by space:").split())
    biggest = get_big(a,b,c)
    print(f"The numbers are:{a},{b},{c} and the biggest is {biggest}")
if __name__ == "__main__":
    main()         
```

```python
#Q.6
def calc_vol(height, length, breadth):
    vol = height * length * breadth
    return vol
    
def main():
    height, length, breadth = map(float, input(" Enter the lenght, width and the breadth of the cube:").split())
    vol = calc_vol(height, length, breadth)
    print(f"Given the dimensions:{height}, {length}, {breadth} ; the volume is : {vol}")
    
if __name__ == "__main__":
    main()          
```

```python
#Q.7
def rectangle_area(length, width):
    return length * width


def circle_area(radius):
    return 22/7 * radius * radius


def triangle_area(base, height):
    return 0.5 * base * height


def main():
    print("Choose a figure")
    print("1. Rectangle")
    print("2. Circle")
    print("3. Right-angled Triangle")

    choice = int(input("Enter your choice (1-3): "))

    if choice == 1:
        length = float(input("Enter length: "))
        width = float(input("Enter width: "))
        area = rectangle_area(length, width)

        print("\nFigure: Rectangle")
        print("Length:", length)
        print("Width:", width)
        print("Area:", area)

    elif choice == 2:
        radius = float(input("Enter radius: "))
        area = circle_area(radius)

        print("\nFigure: Circle")
        print("Radius:", radius)
        print("Area:", area)

    elif choice == 3:
        base = float(input("Enter base: "))
        height = float(input("Enter height: "))
        area = triangle_area(base, height)

        print("\nFigure: Right-angled Triangle")
        print("Base:", base)
        print("Height:", height)
        print("Area:", area)

    else:
        print("Invalid choice")


if __name__ == "__main__":
    main() 
```

```python
#Q.8
def calc_tax(gross_pay):
    if gross_pay > 50000:
        return gross_pay * 0.35
    elif gross_pay >= 40000:
        return gross_pay * 0.30
    elif gross_pay >= 20000:
        return gross_pay * 0.25
    elif gross_pay >= 12000:
        return gross_pay * 0.15
    else:
        return 0

def main():
    gross_pay = float(input("Enter gross pay: "))

    tax = calc_tax(gross_pay)
    net_pay = gross_pay - tax

    print("Gross Pay:", gross_pay)
    print()
    print("Tax Amount:", tax)
    print("Net Pay:", net_pay)

main()
```

```python
#Q.10
def Temp(fahrenheit):
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius

def main():
    print("Fahrenheit\tCelsius")

    for fahrenheit in range(0, 301, 20):
        print(f"{fahrenheit}\t\t{Temp(fahrenheit):.2f}")

main()
```
