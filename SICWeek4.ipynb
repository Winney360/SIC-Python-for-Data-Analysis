```python
#Area and the circumfrence of a circle
pi = 22/7
radius = float(input("Enter the radius"))
area = pi * radius ** 2
circumfrence = 2 * pi * radius 

print(f"The area of the circle with radius {radius}, is {area} and its circumfrence is {circumfrence}")
```

```python
#Area and perimeter of a rectangle 
length = float(input("Enter the length"))
width = float(input("Enter the width"))
area = length * width
perimeter = (2*length)+(2*width)
print(f"The area of a rectangle is :{area} and its perimeter is {perimeter}")
```

```python
#Area of a circle
def fn_area(radius):
    pi = 22/7
    area = pi * radius ** 2
    circumfrence = 2 * pi * radius
    return area,circumfrence

#output
radius = float(input("Enter radius:"))

area,circumfrence = fn_area(radius)

print(f"The area is :{area} and the circumfrence is:{circumfrence)}")


```

```python
#Area of a rectangle
def fn_area(width,length):
    area = width * length
    return area
#Perimeter of a rectangle
def fn_perimeter(width,length):
    perimeter = (2 * width) +(2 * length)
    return perimeter

#output
width =float(input("Enter width:"))
length =float(input("Enter length:"))
print(f"The area of the rectangle is {fn_area(width,length)} and its perimeter is {fn_perimeter(width,length)}")
```

```python
#returning the largest number
def fn_largest(num1,num2,num3):
    if num1 > num2 :
        return num1
    else:
        return num2

#using the function
num1 = float(input("Enter the first number:"))
num2 = float(input("Enter the second number:"))

largest = fn_largest(num1,num2)

print(f"The largest number between the two is :{largest}")
```

```python
def fn_grades(algebra,discrete,statistics):
    average_mark = (algebra + discrete + statistics) / 3
    if average_mark >= 70:
        grade = 'A'
    if average_mark >= 60 and average_mark < 70 :
        grade = 'B'
    if average_mark >= 50 and average_mark < 60 :
        grade = 'C'
    if average_mark >= 40 and average_mark < 50 :
        grade = 'D'
    if average_mark < 40:
        grade = 'E'
    
    return average_mark, grade
    
#use of the function
algebra = float(input("Enter the marks in Algebra: "))
discrete = float(input("Enter the marks in Discrete: "))
statistics = float(input("Enter the marks in Statistics: "))

average_mark, grade = fn_grades(algebra,discrete,statistics)

print(f"The average mark is: {average_mark} and your grade is {grade}." )
```
